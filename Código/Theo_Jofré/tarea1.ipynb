```python
import pandas as pd
import plotly.express as px
from pathlib import Path
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ruta = Path('..') / '..' / 'Fuentes/Studio_Ghibli.csv'
with open(ruta, 'r', encoding='utf-8') as f:
    df = pd.read_csv(f)
```

```python
# Eliminar símbolos "$" y comas, luego convertir a float
df['Budget'] = df['Budget'].str.replace('$', '').str.replace(',', '').astype(float)
df['Revenue'] = df['Revenue'].str.replace('$', '').str.replace(',', '').astype(float)
df['Revenue_Budget_Ratio'] = df['Revenue'] / df['Budget']
fig = px.sunburst(
    df,
    path=['Genre 1', 'Director'],  # Jerarquía: género → director
    values='Budget',               # Tamaño por presupuesto
    color='Revenue_Budget_Ratio',  # Color por ROI
    color_continuous_scale='RdYlGn',
    title='Distribución de Presupuesto y ROI por Género y Director'
)
fig.show()
```
