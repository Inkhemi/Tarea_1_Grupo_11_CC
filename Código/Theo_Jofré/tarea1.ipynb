```python
import pandas as pd
import plotly.express as px
from pathlib import Path
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ruta = Path('..') / '..' / 'Fuentes/Studio_Ghibli.csv'
with open(ruta, 'r', encoding='utf-8') as f:
    df = pd.read_csv(f)
```

```python
# Limpiar columnas monetarias ($ y comas)
df['Budget'] = df['Budget'].str.replace('[\$,]', '', regex=True).astype(float)
df['Revenue'] = df['Revenue'].str.replace('[\$,]', '', regex=True).astype(float)

# Calcular ROI (Revenue/Budget)
df['ROI'] = df['Revenue'] / df['Budget']

# Eliminar filas con valores nulos en columnas clave
df = df.dropna(subset=['Year', 'Director', 'Budget', 'Revenue'])
```

```python
# Crear gráfico de burbujas
fig_bubble = px.scatter(
    df,
    x='Year',
    y='Revenue',
    size='Budget',
    color='Director',
    hover_name='Name',
    title='Recaudación por Año en Studio Ghibli (Tamaño: Presupuesto)',
    labels={'Revenue': 'Recaudación (USD)', 'Year': 'Año'},
    height=600
)

# Personalizar diseño
fig_bubble.update_layout(
    hovermode='closest',
    xaxis=dict(tickmode='linear', tick0=1980, dtick=5),
    showlegend=True
)

# Mostrar gráfico
fig_bubble.show()
```

```python
# Crear treemap
fig_treemap = px.treemap(
    df,
    path=['Director', 'Name'],
    values='Revenue',
    color='ROI',
    color_continuous_scale='RdYlGn',
    title='Recaudación y ROI por Director en Studio Ghibli',
    hover_data=['Year', 'Budget'],
    height=600
)

# Personalizar diseño
fig_treemap.update_traces(
    textinfo='label+value',
    hovertemplate='<b>%{label}</b><br>Revenue: %{value:$,.0f}<br>ROI: %{color:.2f}'
)

# Mostrar gráfico
fig_treemap.show()
```
